# Analyse des patterns de noms

## MySportU
Format: `CODE - NOM EQUIPE`
- CODE: `069069XXX` (numéro d'institution)
- NOM: Nom de l'équipe avec genre (F/M) et numéro

## Config
Format: `NOM (NUMERO)`
- NOM: INSA, LYON 1, LYON 2, etc.
- NUMERO: (1), (2), etc.

## Correspondances à établir
1. Équipes : Matcher le nom et le numéro
2. Gymnases : Mapper les noms courts vers les noms longs MySportU

```python
import re

# Analyser les équipes MySportU pour extraire les composants
def parse_mysportu_team(name):
    """Parse un nom d'équipe MySportU pour extraire le code, l'institution et le numéro"""
    # Format: "069069XXX - NOM EQUIPE"
    match = re.match(r'^(\d+)\s*-\s*(.+)$', name)
    if not match:
        return None, name, None, None
    
    code = match.group(1)
    rest = match.group(2).strip()
    
    # Déterminer le genre (F ou M)
    genre = None
    if 'Féminin' in rest or ' F ' in rest or ' F(' in rest or rest.endswith(' F') or 'F VB' in rest or 'VB F' in rest or 'FILLES' in rest.upper():
        genre = 'F'
    elif 'Masculin' in rest or ' M ' in rest or ' M(' in rest or rest.endswith(' M') or 'M VB' in rest or 'VB M' in rest:
        genre = 'M'
    
    # Extraire le numéro d'équipe
    num_match = re.search(r'[\(\s](\d+)[\)\s]?|(\d+)\s*$|n°(\d+)', rest)
    numero = None
    if num_match:
        numero = num_match.group(1) or num_match.group(2) or num_match.group(3)
    
    return code, rest, genre, numero

# Tester sur quelques équipes
print("Analyse des équipes MySportU:")
for eq in equipes_mysportu[:20]:
    code, name, genre, num = parse_mysportu_team(eq)
    print(f"  {eq}")
    print(f"    -> Code: {code}, Genre: {genre}, Num: {num}")
```

```python
def parse_mysportu_team_v2(name):
    """Parse un nom d'équipe MySportU pour extraire le code, l'institution et le numéro"""
    # Format: "069069XXX - NOM EQUIPE"
    match = re.match(r'^(\d+)\s*-\s*(.+)$', name)
    if not match:
        return None, name, None, None
    
    code = match.group(1)
    rest = match.group(2).strip()
    
    # Déterminer le genre (F ou M)
    genre = None
    rest_upper = rest.upper()
    if 'FÉMININ' in rest_upper or 'FEMININ' in rest_upper or 'FILLES' in rest_upper or ' F ' in rest_upper or ' F(' in rest_upper or rest_upper.endswith(' F') or 'F VB' in rest_upper or 'VB F' in rest_upper:
        genre = 'F'
    elif 'MASCULIN' in rest_upper or ' M ' in rest_upper or ' M(' in rest_upper or rest_upper.endswith(' M') or 'M VB' in rest_upper or 'VB M' in rest_upper:
        genre = 'M'
    
    # Extraire le numéro d'équipe - chercher le pattern (N) ou N à la fin après le genre ou "n°N"
    # Priorité: (N) dans le nom de l'équipe elle-même
    num_match = re.search(r'\((\d+)\)', rest)  # (N)
    if num_match:
        numero = num_match.group(1)
    else:
        # Chercher le numéro à la fin: "VB M 5" ou "n°1"
        num_match = re.search(r'n°(\d+)|VB\s*[MF]\s*(\d+)\s*$|\s(\d+)\s*$', rest)
        numero = None
        if num_match:
            numero = num_match.group(1) or num_match.group(2) or num_match.group(3)
    
    return code, rest, genre, numero

# Tester sur toutes les équipes pour voir les patterns
print("Analyse des équipes MySportU:")
parsed_teams = []
for eq in equipes_mysportu:
    code, name, genre, num = parse_mysportu_team_v2(eq)
    parsed_teams.append({'original': eq, 'code': code, 'name': name, 'genre': genre, 'num': num})
    
df_parsed = pd.DataFrame(parsed_teams)
print(f"\nÉquipes sans genre détecté: {len(df_parsed[df_parsed['genre'].isna()])}")
print(df_parsed[df_parsed['genre'].isna()]['original'].tolist())

print(f"\nÉquipes sans numéro détecté: {len(df_parsed[df_parsed['num'].isna()])}")
print(df_parsed[df_parsed['num'].isna()]['original'].tolist())
```

```python
# Mapping des codes institution vers noms courts
INSTITUTION_CODE_MAP = {
    '069069001': 'ENTPE',   # Attention: doublon avec 069069020
    '069069006': 'LYON 2',
    '069069007': 'LYON 3',
    '069069008': 'ESA',     # EMSLB = ESA Lyon Bron
    '069069009': 'ISARA',
    '069069012': 'CATHO',
    '069069014': 'CPE',
    '069069015': 'ECAM',    # AS = ECAM
    '069069016': 'ECL',     # Centrale Lyon
    '069069018': 'EML',     # EM Lyon
    '069069019': 'ENS',
    '069069020': 'ENTPE',   # ENTPE aussi
    '069069021': 'ESME',    # ESME ?
    '069069022': 'ESSCA',
    '069069025': 'INSA',
    '069069026': 'ISOSTEO',
    '069069028': 'VETO',    # COB = Vétérinaire ?
    '069069029': 'CESI',
    '069069031': 'LYON 1',
}

# Voir les codes uniques dans MySportU
codes_mysportu = df_parsed['code'].dropna().unique()
print("Codes institution MySportU:")
for code in sorted(codes_mysportu):
    eqs = df_parsed[df_parsed['code'] == code]['original'].tolist()
    print(f"\n{code}:")
    for eq in eqs:
        print(f"  {eq}")
```

```python
# Créer un mapping complet MySportU -> Config
# On va créer une table de correspondance basée sur:
# 1. Le code institution
# 2. Le genre (F/M)
# 3. Le numéro d'équipe

# D'abord, construisons le mapping institution
INSTITUTION_CODE_MAP = {
    '069069001': 'ENTPE',   # Attention: équipes masc seulement ici?
    '069069006': 'LYON 2',
    '069069007': 'LYON 3',
    '069069008': 'ESA',     # EMSLB = ESA Lyon Bron  
    '069069009': 'ISARA',
    '069069012': 'CATHO',
    '069069014': 'CPE',
    '069069015': 'ECAM',    # AS = ECAM
    '069069016': 'ECL',     # Centrale Lyon
    '069069018': 'EML',     # EM Lyon
    '069069019': 'ENS',
    '069069020': 'ENTPE',   # ENTPE aussi (fem + masc 2)
    '069069021': 'ESME',    # Volley masc/fem
    '069069022': 'ESSCA',
    '069069025': 'INSA',
    '069069026': 'ISOSTEO',
    '069069028': 'VETO',    # COB = Vétérinaire
    '069069029': 'CESI',
    '069069031': 'LYON 1',  # inclut SANTE (Lyon 1 Santé)
}

def normalize_mysportu_team(name, genre_from_match=None):
    """
    Normalise un nom d'équipe MySportU vers le format de config.
    
    Args:
        name: Nom brut MySportU (ex: "069069025 - AS INSA LYON VOLLEY-BALL MASCULIN 3")
        genre_from_match: Genre déduit du championnat (F/M) si disponible
    
    Returns:
        (config_name, genre, numero): Tuple avec le nom config, le genre et le numéro
    """
    # Extraire le code et le reste
    match = re.match(r'^(\d+)\s*-\s*(.+)$', name)
    if not match:
        return None, None, None
    
    code = match.group(1)
    rest = match.group(2).strip()
    
    # Obtenir l'institution
    institution = INSTITUTION_CODE_MAP.get(code)
    if not institution:
        return None, None, None
    
    # Déterminer le genre
    rest_upper = rest.upper()
    genre = None
    if 'FÉMININ' in rest_upper or 'FEMININ' in rest_upper or 'FILLES' in rest_upper or \
       ' F ' in rest or ' F(' in rest or rest.endswith(' F') or \
       'F VB' in rest_upper or 'VB F' in rest_upper or ' F1' in rest_upper or ' F2' in rest_upper or ' F3' in rest_upper:
        genre = 'F'
    elif 'MASCULIN' in rest_upper or ' M ' in rest or ' M(' in rest or rest.endswith(' M') or \
         'M VB' in rest_upper or 'VB M' in rest_upper or 'VB G' in rest_upper or \
         ') M' in rest_upper or 'M 1' in rest_upper or 'M 2' in rest_upper or 'M 3' in rest_upper:
        genre = 'M'
    
    # Si pas de genre trouvé, utiliser celui du match
    if genre is None and genre_from_match:
        genre = genre_from_match
    
    # Extraire le numéro
    numero = None
    
    # Cas spéciaux selon l'institution
    if institution == 'LYON 2':
        # Format: UDL- UTE LYON 2 VB F (3) ou UDL UTE LYON 2 VB M 5 ou IEP
        if 'IEP' in rest_upper:
            # C'est LYON 2 (IEP) (4)
            num_match = re.search(r'[MF]\s*(\d+)|(\d+)\s*$', rest)
            if num_match:
                numero = num_match.group(1) or num_match.group(2)
            return 'LYON 2 (IEP) (4)', genre, numero or '4'
        num_match = re.search(r'\((\d+)\)|VB\s*[MF]\s*(\d+)', rest)
        if num_match:
            numero = num_match.group(1) or num_match.group(2)
    
    elif institution == 'LYON 1':
        # Vérifier si c'est SANTE
        if 'SANTÉ' in rest_upper or 'SANTE' in rest_upper:
            num_match = re.search(r'\((\d+)\)|Santé\s*\((\d+)\)|Santé(\d+)', rest, re.IGNORECASE)
            if num_match:
                numero = num_match.group(1) or num_match.group(2) or num_match.group(3)
            return f'SANTE ({numero or "1"})', genre, numero
        # Lyon 1 normal
        num_match = re.search(r'\((\d+)\)|LYON\s*1\s*\((\d+)\)', rest, re.IGNORECASE)
        if num_match:
            numero = num_match.group(1) or num_match.group(2)
    
    elif institution == 'INSA':
        num_match = re.search(r'(\d+)\s*$', rest)
        if num_match:
            numero = num_match.group(1)
    
    elif institution == 'ECL':
        # Format: CENTRALE LYON VOLLEY M 1 ou F1/F2
        num_match = re.search(r'[MF]\s*(\d+)|([MF])(\d+)', rest, re.IGNORECASE)
        if num_match:
            numero = num_match.group(1) or num_match.group(3)
    
    elif institution == 'EML':
        # Format: EML 1 F VB
        num_match = re.search(r'EML\s*(\d+)', rest)
        if num_match:
            numero = num_match.group(1)
    
    elif institution == 'ENS':
        # Format: ENS VB F1 ou ENS VB G1 (G = garçon = M)
        num_match = re.search(r'[FG](\d+)', rest)
        if num_match:
            numero = num_match.group(1)
        if 'VB G' in rest_upper:
            genre = 'M'
    
    elif institution == 'ENTPE':
        # Peut être 069069001 ou 069069020
        if 'FILLES' in rest_upper:
            genre = 'F'
            numero = '1'
        else:
            num_match = re.search(r'\((\d+)\)|ENTPE\s*(\d+)', rest)
            if num_match:
                numero = num_match.group(1) or num_match.group(2)
    
    elif institution == 'ESA':
        # EMSLB ou ESA VOLLEY
        num_match = re.search(r'(\d+)\s*$', rest)
        if num_match:
            numero = num_match.group(1)
    
    elif institution == 'ESME':
        # Volley masculin 1/2 ou Volley Féminin 2
        num_match = re.search(r'(\d+)\s*$', rest)
        if num_match:
            numero = num_match.group(1)
    
    elif institution == 'ESSCA':
        numero = '1'  # Une seule équipe par genre
    
    elif institution == 'CPE':
        # CPE ou CPE VB M n°1/2
        num_match = re.search(r'n°(\d+)|\((\d+)\)', rest)
        if num_match:
            numero = num_match.group(1) or num_match.group(2)
        else:
            numero = '1'  # CPE seul = équipe 1
    
    elif institution == 'CATHO':
        num_match = re.search(r'\((\d+)\)', rest)
        if num_match:
            numero = num_match.group(1)
        else:
            numero = '1'  # CATHO seul = équipe 1
    
    else:
        # Cas général: chercher (N) ou N à la fin
        num_match = re.search(r'\((\d+)\)|(\d+)\s*$', rest)
        if num_match:
            numero = num_match.group(1) or num_match.group(2)
    
    if numero:
        config_name = f"{institution} ({numero})"
    else:
        config_name = f"{institution} (1)"  # Default
    
    return config_name, genre, numero

# Tester la fonction sur quelques équipes
test_teams = [
    '069069025 - AS INSA LYON VOLLEY-BALL MASCULIN 3',
    '069069006 - UDL- UTE LYON 2 VB F (3)',
    '069069031 - Lyon 1 Santé (2)',
    '069069007 - LYON3 (2) Volley Féminin',
    '069069016 - CENTRALE LYON VOLLEY M 1',
    '069069018 - EML 2 F VB',
    '069069019 - ENS VB G3',
    '069069020 - ENTPE FIlles',
    '069069014 - CPE VB M n°2',
    '069069022 - Equipe féminine de volley de l\'ESSCA Lyon',
]

print("Test de normalisation:")
for team in test_teams:
    config_name, genre, num = normalize_mysportu_team(team)
    print(f"  {team}")
    print(f"    -> {config_name} ({genre})")
```

```python
# Créer la table de mapping pour toutes les équipes
mapping_results = []

for eq in equipes_mysportu:
    config_name, genre, num = normalize_mysportu_team(eq)
    mapping_results.append({
        'mysportu_name': eq,
        'config_name': config_name,
        'genre_detected': genre,
        'numero': num
    })

df_mapping = pd.DataFrame(mapping_results)

# Comparer avec les équipes de la config
equipes_config_uniques = set()
for _, row in df_equipes_config.iterrows():
    key = (row['Equipe'], row['Genre_Equipe'])
    equipes_config_uniques.add(key)

print(f"Équipes config uniques: {len(equipes_config_uniques)}")
print(f"Équipes MySportU mappées: {len(df_mapping)}")

# Vérifier les mappings qui n'existent pas dans la config
print("\n=== Équipes MySportU dont le mapping n'existe pas dans la config ===")
problemes = []
for _, row in df_mapping.iterrows():
    key = (row['config_name'], row['genre_detected'])
    if key not in equipes_config_uniques and row['config_name'] is not None:
        problemes.append(row)
        print(f"  {row['mysportu_name']} -> {row['config_name']} ({row['genre_detected']})")

print(f"\nTotal problèmes: {len(problemes)}")
```

```python
# Le genre peut être déduit du championnat
# LYON VBF PH2 = Féminin
# LYON VBM PH2 = Masculin

# Créer un mapping équipe -> genre basé sur les matchs
def get_genre_from_championnat(champ):
    if 'VBF' in champ or 'Féminin' in champ:
        return 'F'
    elif 'VBM' in champ or 'Masculin' in champ:
        return 'M'
    return None

# Pour chaque équipe MySportU, trouver le genre via les matchs
equipe_genre_map = {}
for _, row in df_ph2.iterrows():
    genre = get_genre_from_championnat(row['Championnat'])
    if genre:
        equipe_genre_map[row['Equipe A']] = genre
        equipe_genre_map[row['Equipe B']] = genre

print(f"Équipes avec genre déduit du championnat: {len(equipe_genre_map)}")

# Maintenant refaire le mapping avec le genre
mapping_results_v2 = []
for eq in equipes_mysportu:
    genre_from_champ = equipe_genre_map.get(eq)
    config_name, genre_detected, num = normalize_mysportu_team(eq, genre_from_champ)
    final_genre = genre_detected or genre_from_champ
    mapping_results_v2.append({
        'mysportu_name': eq,
        'config_name': config_name,
        'genre': final_genre,
        'numero': num
    })

df_mapping_v2 = pd.DataFrame(mapping_results_v2)

# Vérifier à nouveau
print("\n=== Équipes MySportU dont le mapping n'existe pas dans la config (v2) ===")
problemes_v2 = []
for _, row in df_mapping_v2.iterrows():
    key = (row['config_name'], row['genre'])
    if key not in equipes_config_uniques and row['config_name'] is not None:
        problemes_v2.append(row)
        print(f"  {row['mysportu_name']} -> {row['config_name']} ({row['genre']})")

print(f"\nTotal problèmes: {len(problemes_v2)}")
```

```python
from datetime import datetime, timedelta

# Calculer le numéro de semaine basé sur la date de début de saison
DATE_DEBUT = datetime(2026, 1, 22)  # Depuis config_volleyP2.yaml

def calcul_semaine(date_str):
    """Calcule le numéro de semaine à partir de la date de début."""
    if date_str == 'N-A' or pd.isna(date_str):
        return None
    try:
        date = datetime.strptime(date_str, '%d/%m/%Y')
        # Calculer le nombre de jours depuis le début
        delta = (date - DATE_DEBUT).days
        # Semaine = delta // 7 + 1
        semaine = delta // 7 + 1
        return semaine
    except:
        return None

# Appliquer aux matchs PH2
df_ph2['Semaine_Calculee'] = df_ph2['Date'].apply(calcul_semaine)

print("Distribution des matchs par semaine:")
print(df_ph2['Semaine_Calculee'].value_counts().sort_index())

# Afficher quelques matchs de la semaine 1
print("\n=== Matchs Semaine 1 (22/01/2026) ===")
s1 = df_ph2[df_ph2['Semaine_Calculee'] == 1]
for _, row in s1.iterrows():
    eq_a = normalize_mysportu_team(row['Equipe A'], get_genre_from_championnat(row['Championnat']))[0]
    eq_b = normalize_mysportu_team(row['Equipe B'], get_genre_from_championnat(row['Championnat']))[0]
    genre = get_genre_from_championnat(row['Championnat'])
    gym = GYMNASE_MAP.get(row['Lieu'], row['Lieu'])
    print(f"  {eq_a} vs {eq_b} ({genre}) - {row['Heure']} - {gym}")
```

```python
# Normaliser la colonne Semaine
def normalize_semaine(val):
    if pd.isna(val):
        return None
    if isinstance(val, (int, float)):
        return int(val)
    # Format "N (DD/MM)" -> N
    match = re.match(r'^(\d+)', str(val))
    if match:
        return int(match.group(1))
    return None

df_matchs_fixes['Semaine_Norm'] = df_matchs_fixes['Semaine'].apply(normalize_semaine)

# Comparer semaine 1
print("=== Comparaison Semaine 1 ===")
config_s1 = df_matchs_fixes[df_matchs_fixes['Semaine_Norm'] == 1]
print(f"Matchs fixes Config S1: {len(config_s1)}")

for _, row in config_s1.iterrows():
    print(f"  {row['Equipe_1']} vs {row['Equipe_2']} ({row['Genre']}) - {row['Gymnase']}")

# Créer les clés config
config_s1_keys = set()
for _, row in config_s1.iterrows():
    key = create_match_key(row['Equipe_1'], row['Equipe_2'], row['Genre'])
    config_s1_keys.add(key)

# Comparer avec MySportU
common = mysportu_s1_keys & config_s1_keys
print(f"\nMatchs en commun: {len(common)}")
for key in sorted(common, key=lambda x: str(x)):
    print(f"  {key}")

# Différences
only_mysportu = mysportu_s1_keys - config_s1_keys
only_config = config_s1_keys - mysportu_s1_keys

print(f"\nMatchs MySportU seulement ({len(only_mysportu)}):")
for key in only_mysportu:
    print(f"  {key}")
    
print(f"\nMatchs Config seulement ({len(only_config)}):")
for key in only_config:
    print(f"  {key}")
```
